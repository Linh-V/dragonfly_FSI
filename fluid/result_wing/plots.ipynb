```python
import numpy as np
import matplotlib.pyplot as plt
print("Import completed")
```

**Lift & Drag on a dragonfly wing profile**

Importing the dataset for both $Re = 100$ & $Re = 4000$

```python
Re_100 = np.loadtxt('results_wing_open_Re_100/force_coefficients_channel.txt', delimiter=' ', skiprows=1)
Re_4000 = np.loadtxt('results_wing_open_Re_4000/force_coefficients_channel_restart.txt', delimiter=' ', skiprows=1)

time_100 = Re_100[:, 0]
cd_100 = Re_100[:, 1]
cl_100 = Re_100[:, 2]

time_4000 = Re_4000[:, 0]
cd_4000 = Re_4000[:, 1]
cl_4000 = Re_4000[:, 2]
```

Plotting the $C_l$ & $C_d$ variation with respect to time for $Re = 100$

```python
plt.subplot(2, 1, 1)
plt.plot(time_100, cl_100, label='Cl Re=100')
plt.ylabel('$C_l$')
# plt.title('Aerodynamic Coefficients vs Time, $Re=100$')
plt.ylim(-0.5, 0.5)
plt.grid()

plt.subplot(2, 1, 2)
plt.plot(time_100, cd_100, label='Cd Re=100')
plt.xlabel('Time')
plt.ylabel('$C_d$')
plt.ylim(0, 1)
plt.grid()
```

```python
plt.figure
plt.subplot(2, 1, 1)
plt.plot(time_100, cl_100, label='Cl Re=100')
plt.ylabel('Coefficient of Lift')
plt.title('Aerodynamic Coefficients vs Time - $Re=100$, Zoomed')
plt.ylim(-0.065, -0.055)
plt.xlim(0.04, 0.06)
plt.grid()

plt.subplot(2, 1, 2)
plt.plot(time_100, cl_100, label='Cl Re=100')
plt.xlabel('Time')
plt.ylabel('Coefficient of Lift')
plt.xlim(0.1, 0.12)
plt.ylim(-0.075, -0.065)
plt.grid()
plt.show()

plt.figure()
plt.plot(time_100, cl_100)
plt.legend()
plt.xlabel('Time')
plt.ylabel('Coefficient of Lift')
plt.ylim(-0.072, -0.0715)
plt.xlim(0.30, 0.35)
plt.grid()
plt.show()

plt.figure
plt.subplot(2, 1, 1)
plt.plot(time_100, cd_100, label='Cl Re=100')
plt.ylabel('Coefficient of Lift')
plt.title('Aerodynamic Coefficients vs Time - $Re=100$, Zoomed')
plt.xlim(0.04, 0.06)
plt.grid()

plt.subplot(2, 1, 2)
plt.plot(time_100, cd_100, label='Cl Re=100')
plt.xlabel('Time')
plt.ylabel('Coefficient of Lift')
plt.xlim(0.3, 0.35)
plt.ylim(0, 0.5)
plt.grid()
plt.show()
```

Plotting the $C_l$ & $C_d$ variation with respect to time for $Re = 4000$

```python
plt.subplot(2, 1, 1)
plt.plot(time_4000, cl_4000, label='Cl Re=4000')
plt.ylabel('Coefficient of Lift')
plt.title('Aerodynamic Coefficients vs Time, Re=4000')
plt.ylim(0.2, 0.35)
plt.xlim(0.003, 0.012)
plt.grid()

plt.subplot(2, 1, 2)
plt.plot(time_4000, cd_4000, label='Cd Re=4000')
plt.xlabel('Time')
plt.ylabel('Coefficient of Drag')
plt.xlim(0.003, 0.012)
plt.ylim(0.1, 0.2)
plt.grid()
```

Comparison

```python
plt.subplot(2, 1, 1)
plt.plot(time_100, cl_100, label='Cl Re=100')
plt.plot(time_4000, cl_4000, label='Cl Re=4000')
plt.legend()
plt.xlabel('Time')
plt.ylabel('Coefficient of Lift')
plt.title('Aerodynamic Coefficients vs Time')
plt.xlim(0.003, 0.006)
plt.ylim(1, -1)
plt.legend()
plt.grid()

plt.subplot(2, 1, 2)
plt.plot(time_100, cd_100, label='Cd Re=100')
plt.plot(time_4000, cd_4000, label='Cd Re=4000')
plt.legend()
plt.xlabel('Time')
plt.ylabel('Coefficient of Drag')
plt.xlim(0.003, 0.006)
plt.ylim(1, -1)
plt.legend()
plt.grid()
```
